```python
import zipfile, shutil, os

VOLUME = "/Volumes/workspace/raw/landing"
ORIGEM = f"{VOLUME}/dados_landing.zip"
TEMP = "/tmp/dados_landing"

# 1. Extrai o zip numa pasta temporária do compute
with zipfile.ZipFile(ORIGEM) as z:
    z.extractall(TEMP)

# 2. Copia os arquivos extraídos para o volume
shutil.copytree(TEMP, VOLUME, dirs_exist_ok=True)

# 3. Remove o zip e a pasta temporária
os.remove(ORIGEM)
shutil.rmtree(TEMP)

# 4. Lista o que ficou no volume
for raiz, pastas, arquivos in os.walk(VOLUME):
    for arq in arquivos:
        print(os.path.join(raiz, arq).replace(VOLUME + "/", ""))
```
